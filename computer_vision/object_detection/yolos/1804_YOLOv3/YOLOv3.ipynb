- 如果单元格偏离左上角$(c_x, c_y)$且边界框的先验宽度为$p_w$, 高度为$p_h$，则预测对应：
$$\begin{aligned}
b_x &= \sigma(t_x) + c_x \\
b_y &= \sigma(t_y) + c_y \\
b_w &= p_w e^{t_w} \\
b_h &= p_h e^{t_h} \\
Pr(\text{object}) * IOU(b, \text{object}) &= \sigma(t_o) \\
\end{aligned}$$

- 训练的时候使用SSE(平方误差和)作为损失，则坐标梯度为:$\hat{t}_* - t_*$

- YOLOv3使用逻辑回归预测每个边界框的对象性能分。
  - 如果IOU最高，则置信度标签为1（唯一正样本）。
  - 如果IOU大于阀值(0.5)但不是最佳，则忽略预测（忽略样本）。
  - 如果既不是最佳匹配，也不到IOU阀值，则置信度标签为0，仅计算负样本损失(置信度损失)，不计算坐标损失和类别损失。
