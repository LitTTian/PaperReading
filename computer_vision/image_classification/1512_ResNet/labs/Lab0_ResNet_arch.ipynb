```python
import torch
device = "cuda" if torch.cuda.is_available() else "cpu"
device = torch.device(device)
print(f"Using device: {device}")
```

```python
import torch
import torchvision.models as models
from torchvision.models import ResNet50_Weights
from torchsummary import summary

# ---------------------- 步骤1：加载 ResNet 模型 ----------------------
resnet = models.resnet50(weights=ResNet50_Weights.DEFAULT)  # weights: IMAGENET1K_V1, IMAGENET1K_V2, DEFAULT(V2)

# 若需修改最后一层（如分类任务类别数），可自定义（可选）
# 例如：将默认 1000 类改为 10 类（适合自定义数据集）
num_classes = 10
resnet.fc = torch.nn.Linear(resnet.fc.in_features, num_classes)

# ---------------------- 步骤2：指定设备（CPU/GPU） ----------------------
resnet = resnet.to(device)  # 将模型移到 GPU/CPU

# ---------------------- 步骤3：用 torchsummary 查看结构 ----------------------
# 关键：需传入输入张量的形状（batch_size 可设为 -1 表示任意批量）
# ResNet 输入默认是 (3, H, W)，其中 H/W ≥ 32（通常用 224×224，符合 ImageNet 标准）
summary(
    model=resnet,
    input_size=(3, 224, 224),  # (通道数, 高度, 宽度)，必须匹配模型输入要求
    batch_size=-1,  # 批量大小设为 -1，不限制批量
    device=device.type  # 设备类型（cpu 或 cuda）
)
```
