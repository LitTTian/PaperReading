# LLaVA-NeXT-Interleave：挑战多图像、视频与三维 在大型多模态模型中

## 摘要

视觉指令微调技术在提升大多模态模型（Large Multimodal Models, LMMs）性能方面取得了显著进展。然而，现有开源大多模态模型的研究大多聚焦于单图像任务，其在多图像场景下的应用仍未得到充分探索。此外，以往的大多模态模型研究针对不同场景采用独立的解决方案，导致模型无法跨场景泛化，也难以涌现出新的能力。

针对上述问题，本文提出**LLaVA-NeXT-Interleave**。该模型首次在大多模态模型中实现了对**多图像（Multi-image）、多帧（Multi-frame，即视频）、多视角（Multi-view，即三维）和多补丁（Multi-patch，即单图像）**四大场景的统一建模。为赋予模型上述能力，我们将**交错数据格式**作为通用模板，构建了包含117.76万样本的**M4-Instruct数据集**。该数据集覆盖4个核心领域、14项任务和41个原始数据集。同时，我们构建了**LLaVA-Interleave基准测试集**，用于全面评估大多模态模型的多图像任务性能。

大量实验结果表明，**LLaVA-NeXT-Interleave**模型在多图像、视频和三维任务基准测试中均取得领先性能，同时保持了单图像任务的性能水平。此外，该模型还展现出多项涌现能力，例如跨不同任务设置和模态的任务迁移能力。相关代码已开源至[LLaVA-NeXT项目主页](https://github.com/LLaVA-VL/LLaVA-NeXT)。
