# SDXL: 改进高分辨率图像合成的潜在扩散模型

## 摘要 Abstract

1. 模型改进：相比前代 Stable Diffusion (SD)，SDXL采用了参数量三倍于前者的 UNet 骨干网络，参数扩充的核心原因是加入了更多注意力模块，且借助第二个文本编码器扩大了交叉注意力的上下文范围。
2. 技术设计：设计了多种新型条件控制方案，并基于多宽高比开展模型训练；同时引入优化模型，通过事后的图像到图像技术，提升SDXL生成样本的视觉真实度。
3. 性能表现：与前代SD相比性能大幅提升，生成效果可与当前黑盒技术路线的顶尖图像生成模型比肩。
4. 开源举措：为推动开放研究、提升大模型训练与评估的透明度，该研究公开了模型代码与权重。

> #### Algorithm 1 Conditioning pipeline for size- and crop-conditioning
> ---
> - **Require**: Training dataset of images $\mathcal{D}$, target image size $s = (h_{tgt}, w_{tgt})$ <br>
> - **Require**: Resizing function $R$, cropping function function $\mathcal{C}$ <br>
> - **Require**: Model train step $T$ <br>
>   - converged $\leftarrow$ False <br>
>   - **while** not converged **do** <br>
>      - $x \sim \mathcal{D}$ <br>
>      - $w_\text{original} \leftarrow \text{width}(x)$ <br>
>      - $h_\text{original} \leftarrow \text{height}(x)$ <br>
>      - $c_\text{size} \leftarrow (w_\text{original}, h_\text{original})$ <br>
>      - $x \leftarrow R(x, s)$ resize smaller image size to target size **$s$** <br>
>   - **if** $h_\text{original} \le w_\text{original}$ **then** <br>
>      - $c_\text{left} \sim \mathcal{U}(0, \text{width}(x) - s_w)$  sample $c_\text{left}$ from discrete uniform distribution<br>
>      - $c_\text{top} = 0$ <br>
>   - **else if** $h_\text{original} > w_\text{original}$ **then** <br>
>      - $c_\text{top} = \mathcal{U}(0, \text{height}(x) - s_h)$ sample $c_\text{top}$ from discrete uniform distribution<br>
>      - $c_\text{left} = 0$ <br>
>   - **end if** <br>
>      - $c_\text{crop} \leftarrow (c_\text{top}, c_\text{left})$ <br>
>      - $x \leftarrow \mathcal{C}(x, s, c_\text{crop})$ crop image to size s with top-left coordinate ($c_\text{top}, c_\text{left}$) <br>
>      - $\text{converged} \leftarrow T(x, c_\text{size}, c_\text{crop})$ train model conditioned on $c_\text{size}$ and $c_\text{crop}$ <br>
> - **end while**
