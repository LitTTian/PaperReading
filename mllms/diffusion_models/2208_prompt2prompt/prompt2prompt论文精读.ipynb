__图3 方法概述__
1. 上半部分：视觉嵌入与文本嵌入通过交叉注意力层实现融合，为每个文本标记生成对应的空间注意力图。
2. 下半部分：我们利用源图像的注意力图，控制生成图像的空间布局与几何形态。这一设计支持仅通过编辑文本提示词完成多种编辑任务：

- 当替换提示词中的某个词汇时，我们注入源图像的注意力图$M_t$，以此覆盖目标图像的注意力图 $M_t^*$，从而保留原有的空间布局；
- 当新增文本短语时，仅注入与提示词中未改动部分对应的注意力图；
- 通过对目标词汇对应的注意力图进行权重调整，即可增强或弱化该词汇的语义表达效果。
