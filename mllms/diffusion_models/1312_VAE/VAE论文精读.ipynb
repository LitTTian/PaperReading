# 自编码变分贝叶斯

## 摘要

在存在连续潜变量且后验分布难以处理以及大型数据集的情况下，我们如何进行高效的推理和学习？我们引入了一种随机变分推理和学习算法，该算法可扩展到大型数据集，并且在某些温和的可微条件甚至适用于难以处理的情况。
我们的贡献有两个方面。
- 首先，我们证明了变分下界(variational lower bound)的重新参数化(reparameterization)可以得到一个可以用标准随机梯度方法(standard stochastic gradient methods)直接优化的下界估计器。
- 其次，我们证明了对于每个数据点具有连续潜变量的独立同分布数据集，通过使用所提出的下界估计器将近似推理模型（也称为识别模型）拟合到难以处理的后验分布，后验推理可以特别高效。理论优势体现在实验结果中。

## 方法

本节中的策略可用于为具有连续潜变量的各种有向图模型推导一个下界估计器（随机目标函数）。在这里，我们将自己限制在常见情况下，即我们有一个具有每个数据点潜变量的独立同分布数据集，并且我们希望在（全局）参数上执行最大似然（ML）或最大后验（MAP）推理，并在潜变量上进行变分推理。例如，将此场景扩展到我们也在全局参数上进行变分推理的情况是直接的；该算法放在附录中，但这种情况的实验留待未来工作。请注意，我们的方法可以应用于在线、非平稳设置，例如流数据，但在这里我们假设一个固定数据集以简化问题。
