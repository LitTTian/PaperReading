# 学会解释：用于科学问答的多模态推理思维链

- [论文](https://arxiv.org/abs/2209.09513)
- [GitHub](https://scienceqa.github.io/)

## 摘要

__研究背景__
人类回答问题时，会整合多模态信息，形成连贯完整的**思维链（CoT）**。但深度学习模型（如大语言模型）的这一过程通常是黑箱。现有科学问答基准虽用于评估AI的多步推理能力和可解释性，但存在**缺陷**：缺乏答案标注、仅限文本模态、规模小、领域多样性不足。

__研究贡献__
为此，我们提出多模态科学问答基准**ScienceQA**：包含约2.1万道多模态选择题，覆盖多样科学主题，且每道题均配有**课程讲解、答案及解析标注**。

__实验结果__
1.  我们设计模型，通过生成**课程讲解和解析作为思维链**，模拟人类多步推理过程；
2.  思维链提升了模型性能：小样本GPT-3准确率提升1.20%，微调后的UnifiedQA提升3.99%；
3.  将解析直接输入模型时，小样本GPT-3性能**大幅提升18.96%**（模型性能上限实验）；
4.  分析表明：与人类类似，语言模型借助解析可实现**少数据学习**——仅用40%的数据即可达到原模型性能。

## 3 数据集（ScienceQA）

__1. 数据集概述__
我们构建了**ScienceQA**——一个包含21,208道多模态科学选择题的数据集。任务要求模型根据科学问题和多模态上下文，从选项中选出正确答案。

与现有数据集相比，ScienceQA的核心优势：
- **跨领域覆盖**：涵盖自然科学、社会科学、语言科学三大领域；(natural science, social science, and language science)
- **丰富标注**：83.9%的题目配有**课程讲解**（通用背景知识），90.5%的题目配有**答案解析**（具体推理过程）；
- **多模态特性**：支持文本、图像（自然图像/示意图）、多模态混合等上下文形式；
- **核心目标**：推动可解释模型的研发，要求模型生成连贯的思维链，复现多步推理过程。

### 3.1 数据分析
__核心数据__
- **数据划分**：按60:20:20随机分为训练集（12,726题）、验证集（4,241题）、测试集（4,241题）；
- **核心统计**（表1）：
  - 独立问题9,122道，独立课程讲解261篇；
  - 上下文分布：图像上下文48.7%、文本上下文48.2%、混合上下文30.8%、无上下文33.9%；
  - 层级分类：26个主题、127个类别、379个技能点；
  - 文本长度：问题平均12.11词，选项平均4.40词，课程讲解平均125.06词，解析平均47.66词。

__问题分析__
- 问题长度跨度大（3-141词），平均12.11词；
- 与其他视觉问答（VQA）数据集相比，**问题长度分布更均匀**（图1a）；
- 问题文本的前四个词分布多样（图1b），体现题目形式的丰富性。

__上下文分析__
- 66.11%的题目含至少一种上下文（图像/文本），30.8%含混合上下文；
- 图像上下文包含7,803个独立图像（自然图像/示意图），文本上下文包含4,651个独立文本；
- 要求模型具备**灵活的多模态理解能力**，以应对不同类型的上下文输入。

__领域多样性__
- 题目分属自然科学、社会科学、语言科学三大领域；
- 按“主题→类别→技能点”三级分类，共26个主题、127个类别、379个技能点；
- 领域分布均匀，覆盖范围广（图2）。

### 3.2 与现有数据集的对比
ScienceQA与其他科学问答数据集的核心差异（表2）：
- **规模与模态**：规模大（21,208题），图像数量最多（10,332张），支持多模态上下文；
- **覆盖范围**：唯一覆盖1-12年级、三大科学领域的数据集；
- **标注完整性**：首个大规模多模态数据集，同时提供课程讲解（83.9%）和答案解析（90.5%）；
- **问题特性**：问题最长（最大141词），平均长度12.11词，输入源多样性最高。
