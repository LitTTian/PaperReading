````python
import json
from pprint import pprint
from PIL import Image
from collections import defaultdict
with open('./dataset/SpatialScore.json', 'r', encoding='utf-8') as f:
    spatial_data = json.load(f)
with open('./dataset/VGBench.json', 'r', encoding='utf-8') as f:
    vgbench_data = json.load(f)
print(len(spatial_data))
print(len(vgbench_data))

def display_data(data):
    img_paths = data['img_paths']
    imgs = [Image.open(img_path) for img_path in img_paths]
    for img in imgs:
        display(img)
    pprint(data)


def data_analysis(data_list):
    # 统计list中每个字典的value
    unique_values = defaultdict(set)
    for data in data_list:
        for key, value in data.items():
            if isinstance(value, list):
                unique_values[key].update(value)
            else:
                unique_values[key].add(value)
    # 转换为普通字典并将集合转换为列表
    result = {key: list(values) for key, values in unique_values.items()}
    return result

def show_result(analysis_result, threshold=70):
    for key, values in analysis_result.items():
        if len(values) < threshold:
            print(f"{key}: {len(values)} : {values}")
        else:
            print(f"{key}: {len(values)} unique values")
````

````python
show_result(data_analysis(spatial_data))
print("\n" + "="*50 + "\n")
show_result(data_analysis(vgbench_data))
````
