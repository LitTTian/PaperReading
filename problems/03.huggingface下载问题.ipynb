有的时候会遇到huggingface的下载网络问题，我们可以使用镜像站下载

```shell
# 首先安装huggingface-hub和hf_transfer
pip install -U huggingface-hub hf_transfer
export HUGGING_FACE_HUB_TOKEN="your_huggingface_token"
export HF_ENDPOINT="https://hf-mirror.com"
hf download lmms-lab/llava-onevision-qwen2-7b-ov --local-dir ~/Study/models/llava/llava-onevision-qwen2-7b-ov
```
