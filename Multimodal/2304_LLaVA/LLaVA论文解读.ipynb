# 视觉指令微调

TL;DR:

- 动机：将指令微调扩展到视觉-语言多模态领域，构建一个能够理解视觉信息并遵循指令的模型。
  - 利用GPT-4生成数据的方法。
- LLaVA 模型：结合了 CLIP的视觉编码器和 Vicuna的语言解码器，他们之间设计一个线性投影层$W$，将视觉特征映射到语言模型的词嵌入空间，实现了视觉信息与语言指令的融合。
- 两阶段训练：
  - 阶段一：特征对齐预训练，将视觉特征与语言模型的词嵌入空间对齐，只更新投影层
  - 阶段二：端到端微调，在GPT-4生成的视觉指令遵循数据上训练，更新投影层和语言模型的参数，视觉编码器仍保持冻结。
- 性能表现：在多模态对话和科学问题解答任务上，LLaVA 取得了优异的性能，证明了方法的有效性。
- 开源：作者公开了生成的视觉指令数据、代码库、模型检查点以及视觉聊天演示，促进了后续研究。

## 一、背景介绍

在人工智能领域，大语言模型（LLMs） 的指令微调（Instruction Tuning）已经在文本领域取得了巨大的成功，例如 ChatGPT 和 GPT-4 能够理解并执行复杂的人类指令。然而，在多模态（视觉-语言） 领域，指令微调的研究还相对缺乏。

人类的认知是多模态的，我们通过视觉、语言等多种感官来理解世界。因此，构建一个能够同时理解视觉和语言指令的通用视觉助手，对于推动人工智能的发展具有重要意义。

## 三、GPT 辅助的视觉指令数据生成

### 3.1 数据缺乏问题

__多模态指令遵循数据稀缺__，主要原因是数据的标注过程耗时且难以定义。需要大量的人力来为图像生成高质量的指令和回答。

### 3.2 数据生成方法

为了解决这一问题，本文提出了利用 GPT-4 生成视觉指令数据的方法，具体步骤如下：

步骤1：符号表示转换：将图像转化为 GPT 可理解的符号表示，包括：

- 图像字幕（Captions）：描述视觉场景的文本。
- 边界框（Bounding Boxes）：定位图像中的对象及其空间位置。

步骤2：提示 GPT-4 生成数据：使用上述符号表示，提示 GPT-4 生成包含丰富多样的指令和回答。

### 3.3 指令遵循数据的类型

生成了三种类型的指令遵循数据：

- 对话型（Conversation）：模拟人与助手之间的对话，就图像内容提出问题，包含多轮交互。
- 详细描述型（Detailed Description）：要求助手对图像进行详细、全面的描述，涵盖各种细节。
- 复杂推理型（Complex Reasoning）：提出需要深入推理的问题，要求助手进行逻辑严谨的回答。

## 四、视觉指令微调方法

### 4.1 模型架构

LLaVA（Large Language and Vision Assistant） 的架构由两部分组成：

- 视觉编码器：选用预训练的 CLIP ViT-L/14，用于提取图像特征$Z_v = g(X_v)$。
- 语言解码器：选用 Vicuna，一种强大的开源语言模型，具有出色的指令遵循能力。
- 特征映射：设计一个线性投影层$W$，将视觉特征映射到语言模型的词嵌入空间，即$H_v=W\cdot Z_v$。

### 4.2 训练策略

- 对于每张图片$X_v$, 生成了多轮对话数据$(X_q^1, X_a^1, \ldots, X_q^T, X_a^T)$
$$\begin{aligned}
X_\text{instruct}^t = \begin{cases}
\text{Randomly choose} [X_q^1, X_v] \text{or} [X_v, X_q^1] & t=1 \\
X_q^t, & t > 1
\end{cases}
\end{aligned}$$

训练过程分为两个阶段：

__阶段一：特征对齐的预训练__ (更新$W$)

- 目标：将视觉特征与语言模型的词嵌入空间对齐。
- 方法：在过滤后的 595,000 条图像-文本对数据上训练，仅更新投影层$W$，冻结视觉编码器和语言模型的参数。
- 损失函数：最大化模型在生成文本序列的似然概率，公式如下：
$$\theta = \arg\max_\theta \prod_{i=1}^{L} P(x_i|x_{<i}, X_{\text{instruct}}, H_v; \theta)$$
其中$x_i$为第$i$个目标词汇，$X_{\text{instruct}}$为指令，$H_v$为视觉嵌入。

__阶段二：端到端微调__ (更新$W$和语言模型参数)


- 目标：进一步提升模型的指令跟随能力。
- 方法：在生成的 158,000 条视觉指令遵循数据上训练，更新投影层和语言模型的参数，视觉编码器仍保持冻结。
- 应用场景：
  - 多模态聊天机器人：训练模型在多轮对话中理解并响应视觉相关的指令。
  - ScienceQA 数据集：微调模型以回答科学领域的多模态问题。

## 五、实验与结果

### 5.1 多模态聊天机器人

__模型能力展示__
- 案例对比：与 BLIP-2 和 OpenFlamingo 等模型相比，LLaVA 能够更准确地理解图像，并按照指令生成符合要求的回答。
- 示例：
用户输入一张人在汽车顶上熨衣服的图像，提问：“这张图片中有什么不寻常的地方？”
  - LLaVA 回答：详细描述了人在汽车顶上熨衣服的场景，指出了这是一种不寻常的行为。
  - 其他模型：要么只简单描述图像内容，要么未能按照指令回答。


__性能评估__

- LLaVA-Bench（COCO）：
  - 使用 COCO-Val-2014 数据集中的 30 张图像，每张图像生成三类问题，共 90 个样本。
  - 评估模型在不同类型问题上的表现，LLaVA 相对于 GPT-4(纯文本) 达到了 85.1% 的相对得分。

- LLaVA-Bench（In-the-Wild）：
  - 选取 24 张多样化的实际生活图像，提出具有挑战性的 60个 问题。(室内外场景、表情包、绘画、速写等)
  - LLaVA 的整体得分为 67.3%，显著优于其他模型。

__模型局限性__
- 在处理高分辨率图像或需要深入专业知识的问题时，LLaVA 可能表现不足。
- 对于某些细节丰富或超出训练数据分布的场景，模型的理解能力有限。

## 六、QA

1. 为什么使用 GPT-4 生成数据而不是人工标注？
- 由于人工标注大量的视觉指令遵循数据耗时且成本高昂。使用 GPT-4，可以高效生成高质量的数据，早期实验也表明 GPT-4 生成的数据质量优于其他模型。

2. 模型如何将图像信息传递给仅接受文本输入的 GPT-4？
- 通过将图像转换为 GPT-4 可理解的符号表示，如图像的描述性字幕和边界框信息，将视觉信息以文本形式输入 GPT-4。

3. LLaVA 可以处理哪些类型的视觉任务？
- LLaVA 在多模态对话、详细描述和复杂推理等任务上表现出色，能够理解并响应涉及视觉内容的各种指令。

4. 为什么阶段一特征对齐预训练时，只更新投影层？
- 保留预训练模型的知识：保持视觉编码器和语言模型的参数不变，保留其在大规模数据上学到的能力。
- 高效学习映射函数：通过训练投影层，学习将视觉特征映射到语言嵌入空间的函数，实现特征对齐。
- 提高训练效率和稳定性：减少需要更新的参数数量，加快训练速度，避免复杂的训练过程。
- 为后续微调做准备：在完成特征对齐后，再进行端到端微调，可以更有效地训练模型，实现更好的性能。

5. 为什么阶段二端到端微调时，视觉编码器仍保持冻结呢？
- 保留预训练的通用视觉特征：视觉编码器（例如 CLIP 的视觉编码器）已经在大规模的数据集上进行了预训练，学到了丰富且通用的视觉特征表示。这些特征对于各种视觉任务都具有很好的泛化能力。通过冻结视觉编码器，我们可以保留这些预训练的特征，不会因为在相对较小的指令遵循数据集上微调而导致其泛化能力下降。
- 避免过拟合：指令遵循训练数据（即阶段二的数据集）相对于预训练视觉编码器所使用的数据集要小得多。如果在这么小的数据集上更新视觉编码器的参数，可能会导致模型过拟合于新数据，无法很好地泛化到未见过的图像。
提高训练效率：冻结视觉编码器可以显著减少需要更新的参数数量。这不仅降低了训练的计算成本和内存需求，还加快了训练速度。
- 聚焦于语言模型的适应：阶段二的主要目标是让语言模型能够有效地利用视觉特征来生成符合指令的响应。通过冻结视觉编码器，模型的优化重点就放在了语言模型和投影层上，确保语言模型更好地理解和处理来自视觉编码器的特征。

### 补充问题


为什么LLaVA不更新视觉编码器的参数？

1. 极高的计算开销
2. 防止灾难性遗忘和过拟合

visual instruction tuning和visual prompt tuning的区别是什么？

1. visual instruction tuning 提升模型对指令的理解和响应能力
2. visual prompt tuning (PEFT) 旨在通过少量参数调整，使模型适应特定任务或领域
