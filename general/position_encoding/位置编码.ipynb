## 位置编码常见分类

### 1. 绝对位置编码(Absolute Positional Encoding):
每个位置对应一个唯一向量，直接加到输入嵌入上。
1. 固定正弦：([Transformer]()，不可学习，外推性略好)
$$PE_{(pos, 2i)} = sin(\frac{pos}{10000^{\frac{2i}{d_{model}}}}) \\ 
PE_{(pos, 2i+1)} = cos(\frac{pos}{10000^{\frac{2i}{d_{model}}}})$$
1. 可学习参数：（ViT，BERT，GPT可以学习优化，对任务适配性更强）

### 2. 相对位置编码(Relative Positional Encoding):

不编码绝对位置，而是编码token间的相对距离（如偏移量$i-j$），通常在自注意力注入相对偏置，不是在输入嵌入中添加位置编码。核心优势是平移不变性，外推长序列更稳定。
1. 原始相对PE：Transformer-XL、XLNet，在注意力分数中添加相对位置偏置。
2. 窗口相对偏置：Swin Transformer，窗口内计算注意力时加入相对位置偏置，优化局部空间建模
3. 旋转位置编码（RoPE）：LLaMA、GPT-4 等，通过旋转矩阵隐式编码相对位置，外推性极强
4. ALiBi：无位置向量，直接在注意力分数中减 “距离 * 系数”，极简高效

### 3. 条件位置编码(Conditional Positional Encoding):
位置向量依赖输入内容动态生成，而非固定参数或函数；通常通过轻量卷积/池化提取局部邻域特征，生成适配当前输入的位置编码，提升对分辨率、尺度的泛化性与平移不变性。
1. CPVT：去掉ViT中的APE，用局部卷积生成条件位置向量

### 4. 无位置编码(Position Encoding Free):
完全不注入显式位置信息，依赖模型自身从token内容或结构中隐式学习位置关系。
1. 一般作为对比实验基线，验证位置编码的重要性
